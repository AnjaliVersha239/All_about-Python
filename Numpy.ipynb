### Basic Array Creation

```python
# Numpy array is always created using a list
# A List can have multiple data types
a = [1,2,3,4.5,"Hello"]
print(a)
b = [1,2,3,4.5]
print(b)
# A Numpy array
arr = np.array([1,2,3,4])
arr
```

```python
'''If we pass a List in an array that have a data type other than integer it will convert all the data in that perticular data type, if it has a
string will convert all the data in strings using unicodes and if it has a float it will convert all the data present inside the list to float'''
arr1 = np.array(a)
print(arr1)
# If it has a float it will convert all the data into float
arr2 = np.array(b)
print(arr2)
```

```python
''' Vectors are something that is created in single one dimensional space, in simple word a list is a vector but a matrix is 2-dimensional that has 
multiple rows and colunms. we can create a matrix by creating an array inside another array '''
l = [[1,2,3],[4,5,6],[7,8,9]]
print(l)
mat = np.array(l)
print(mat)
```

### Array Generation Function

```python
'''To generate something the first thing we can do is to use range(arange) function same as python we have to give start stop and step value'''
arr3 = np.arange(1,11)
```

```python
'''We can generate zeros and ones for both vector and matrix using Numpy'''
# To generate zeros in vector
arr4 = np.zeros(5)
print(arr4)
# To generate ones in matrix we have to specify the number of rows and columns
arr5 = np.ones((6,8))
print(arr5)
```

```python
'''Linspace Returns evenly spaced numbers over a specified interval we have to give start,stop values and number of elements we want to have'''
arr6 = np.linspace(1,5,4) # It will give 4 elements between 1-5
arr6
```

### Reshaping and Resizing 
As understood by name reshaping means changing the shape of an array, if we have a vector we can change the shape of it in a 2-d matrix using the reshape function we just have to make sure that the number of rows*coulmns should be equal to number of total element in the array and save it in the shape variable to resize it

```python
ara = np.arange(1,31)
ara
```

#### Broadcasting
If we want to do some operation with the entire array and don't want to do it separately with a Loop there we can use broadcasting.

```python
l = [10,20,30,40]
arr12 = np.array(l)
arr12
```

```python
# We can add 10 to each element of the array arr12 at once
arr12 + 10
```

```python
arr13 = np.arange(1,26).reshape(5,5)
arr13
```

```python
arr13 * 2
```

### Matrix Operations

```python
mat1 = np.array([[1,2],[3,4]])
mat2 = np.array([[5,6],[7,8]])
mat1
```
