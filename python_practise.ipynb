```python
#Write a programe to calculate percentage of a student through 5 subject. Take marks as input from the user. using percentage print which grade the student has scored.
maths = int(input("Enter the marks of Maths: "))
bio = int(input("Enter the marks of Biology: "))
chem = int(input("Enter the marks of chemistry: "))
phy = int(input("Enter the marks of physics: "))
eng = int(input("Enter the marks of english: "))
sub_marks = maths+bio+chem+phy+eng
Total_marks = int(input("Enter total marks of the subjects: "))
perc = (sub_marks/ Total_marks)*100
print(perc)
if perc>90:
    print("You got A+ Grade")
elif perc<90 and perc>=80:
    print("You got A Grade")
elif perc <80 and perc>=70:
    print("You got B+ Grade")
elif perc<70 and perc>=60:
    print("You got B Grade")
elif perc<60 and perc>=50:
    print("You got C+ Grade")
elif perc<50 and perc>=40:
    print("You got C Grade")
elif perc<40 and perc>=33:
    print("Your got D Grade")
else:
    print("You are Fail")
```

```python
# Write a python function to rverse a string in python
def rev_str(s):
    return s[::-1]
print(rev_str ("hello"))
```

```python
# Write a python function to check if the string is palindrome
def is_palindrome(a):
    return a == a[::-1]
print(is_palindrome ("anjali"))
```

```python
# Accept a number and check if it is a prefect number or not.
k = int(input("Enter your number: "))
sum = 0
for i in range(1,k):
    if k%i == 0:
        sum = sum + i
if sum == k:
    print("The number is perfect")
else:
    print("The number is not perfect")
```

```python
# Accept a number from the user and check if it is prime or not
l = int(input("Enter your number: "))
count = 0
for i in range(1,l+1):
    if l%i == 0:
        count = count + 1
if count == 2:
    print("The number is prime")
else:
    print("The number is not prime")  
```
